### Parte espacial de la métrica FLRW

```python
import einsteinpy.symbolic as es
import sympy as sp
import numpy as np
```

```python
c, t, r, theta, phi, K = sp.symbols("c t r theta phi, K")

# métrica espacial
gesp_mat = np.diag([1/(1-K*r**2), r**2, r**2*sp.sin(theta)**2])

# métrica como objeto de EinsteinPy
gesp = es.MetricTensor(gesp_mat, [r, theta, phi])

gesp.tensor()
```

Para saber si hay curvatura debo revisar el tensor de Riemann:

```python
R = es.RiemannCurvatureTensor.from_metric(gesp)
R.tensor()
```

```python
Resc = es.RicciScalar.from_metric(gesp)
Resc.tensor()
```

## El espacio-tiempo de Weyl al completo

```python
a=sp.Function("a")

# métrica espacial
gmat = np.diag([c**2, -a(t)**2*1/(1-K*r**2), -a(t)**2*r**2, -a(t)**2*r**2*sp.sin(theta)**2])

# métrica como objeto de EinsteinPy
g = es.MetricTensor(gmat, [t, r, theta, phi])

g.tensor()
```

```python
Resc = es.RicciScalar.from_metric(g)
Resc.tensor().subs(K, 0)
```

```python
G = es.EinsteinTensor.from_metric(g)
G.tensor()
```
